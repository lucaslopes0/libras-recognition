# 📊 Exploração do Dataset V-LIBRASIL

Use após `scripts/02_preprocess.py` para entender o dataset antes do treino.

```python
import numpy as np
import matplotlib.pyplot as plt
from libras.config import PATHS
from libras.data.loader import list_classes
```

```python
classes = list_classes(min_samples=1)
print(f'Total de classes: {len(classes)}')
print(f'Primeiras 20: {classes[:20]}')
```

```python
# Distribuição de amostras
counts = [
    len(list((PATHS['data_processed'] / c).glob('*.npy')))
    for c in classes
]
print(f'Min: {min(counts)} | Max: {max(counts)} | Média: {np.mean(counts):.1f}')

plt.figure(figsize=(12, 4))
plt.bar(range(len(classes)), sorted(counts, reverse=True), color='steelblue')
plt.axhline(3, color='red', linestyle='--', label='Mínimo recomendado (3)')
plt.title('Amostras por classe (ordenado)')
plt.xlabel('Classes'); plt.ylabel('Número de vídeos')
plt.legend(); plt.tight_layout(); plt.show()
```

```python
# Heatmap dos keypoints de um exemplo
example_class = classes[0]
example_file = next((PATHS['data_processed'] / example_class).glob('*.npy'))
arr = np.load(example_file)

print(f'Classe: {example_class}')
print(f'Shape: {arr.shape}  (frames × keypoints)')

plt.figure(figsize=(12, 4))
plt.imshow(arr.T, aspect='auto', cmap='viridis')
plt.colorbar(label='Valor')
plt.title(f'Heatmap de keypoints — {example_class}')
plt.xlabel('Frame'); plt.ylabel('Dimensão')
plt.tight_layout(); plt.show()
```
